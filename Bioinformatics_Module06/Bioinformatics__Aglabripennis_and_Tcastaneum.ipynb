```python
#All comments are mine
#Code generated by chat gpt following the below outline
##Load files
##Define dictionaries to store best blast matches
##Loop through each iteration of agla and tcas and identify the gene and its correlating hit
##check for reciprocal matches and print results

#Load the files
agla_file = "/content/Agla_query_v_Tcas_subject.txt"
tcas_file = "/content/Tcas_query_v_Agla_subject.txt"

#Create "dictionaries" that will store the BLAST matches
agla_to_tcas = {} # A. glabripennis as the query and as the T. castaneum subject
tcas_to_agla = {} # T. castaneum as the query and as the A. glabripennis subject

#Loops through each line
with open(agla_file) as f:
    for line in f:
        qseqid, sseqid, evalue = line.strip().split() #Divides the BLAST into 3 columns for qseqid, sseqid, evalue
        agla_to_tcas[qseqid] = sseqid #stores the agla gene and the best tcas hit

#Same as above
with open(tcas_file) as f:
    for line in f:
        qseqid, sseqid, evalue = line.strip().split()
        tcas_to_agla[qseqid] = sseqid #stores the tcas gene and the best agla hit

#Creates a place to store the RBBH info
rbbh = []

#Checks for recriprocal matches
for agla_gene, tcas_gene in agla_to_tcas.items(): #identifies the best agla and tcas best hit pairs
    if tcas_to_agla.get(tcas_gene) == agla_gene: #checks if hit is recriprocal
        rbbh.append((agla_gene, tcas_gene)) #saves it

print("Number of 2-way orthologous pairs:", len(rbbh))
```
