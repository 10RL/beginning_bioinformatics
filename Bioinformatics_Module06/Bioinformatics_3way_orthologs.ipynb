```python
#All comments are mine
#Code generated by chat gpt following the below outline
##Load files
##2-way analysis
###Define dictionaries to store best blast matches
###Loop through each iteration of agla and tcas and identify the gene and its correlating hit
###check for reciprocal matches and print results
##Calculate all three sets of two way analysis
##3-way analysis
###define dictionaries to store information
###compare tcas and ldec to agla, then compare tcas and ldec to each other
###save results and print

#Load the files
##In the name, the query is first and the subject is second
agla_tcas = "/content/Agla_query_v_Tcas_subject.txt"
tcas_agla = "/content/Tcas_query_v_Agla_subject.txt"

agla_ldec = "/content/Agla_query_v_Ldec_subject.txt"
ldec_agla = "/content/Ldec_query_v_Agla_subject.txt"

tcas_ldec = "/content/Tcas_query_v_Ldec_subject.txt"
ldec_tcas = "/content/Ldec_query_v_Tcas_subject.txt"


#Identify the reciprocal best BLAST hits
def get_rbh(file1, file2): #creates the function and tells it to consider the original direction (File 1) and the reciprical direction (File 2)

    hits1 = {} #creates a dictionary to store blast from file 1
    with open(file1) as f:
      for line in f:
        qseqid, sseqid, evalue = line.strip().split() #Divides the BLAST into 3 columns for qseqid, sseqid, evalue
        if qseqid not in hits1:
            hits1[qseqid] = sseqid #stores the file 1 gene and the best file 2 hit


    hits2 = {} #creates a dictionary to store blast from file 2
    with open(file2) as f:
      for line in f:
          qseqid, sseqid, evalue = line.strip().split() #Divides the BLAST into 3 columns for qseqid, sseqid, evalue
          if qseqid not in hits2:
            hits2[qseqid] = sseqid #stores the file 2 gene and the best file 1 hit

    #Only keep the matches that are reciprical
    rbh = set() #creates a way to store the best matches of the recipricals
    for qseqid, sseqid, in hits1.items(): #identifies the best hit pairs
        if sseqid in hits2 and hits2[sseqid] == qseqid: #checks is the hits are reciprocal
            rbh.add((qseqid, sseqid)) #saves it

    return rbh


#Calculate all 3 sets of 2-way RBBHs
agla_tcas_rbh = get_rbh(agla_tcas, tcas_agla)
agla_ldec_rbh = get_rbh(agla_ldec, ldec_agla)
tcas_ldec_rbh = get_rbh(tcas_ldec, ldec_tcas)

print("Agla-Tcas 2-way RBBHs:", len(agla_tcas_rbh))
print("Agla-Ldec 2-way RBBHs:", len(agla_ldec_rbh))
print("Tcas-Ldec 2-way RBBHs:", len(tcas_ldec_rbh))


#Create dictionaries to store information about each gene
agla_to_tcas = dict(agla_tcas_rbh) #agla to tcas
tcas_to_agla = {t: a for a, t in agla_tcas_rbh} #reciprical

agla_to_ldec = dict(agla_ldec_rbh)
ldec_to_agla = {l: a for a, l in agla_ldec_rbh}

tcas_to_ldec = dict(tcas_ldec_rbh)
ldec_to_tcas = {l: t for t, l in tcas_ldec_rbh}


#Find 3-way reciprocal orthologs
three_way = [] #creates an empty list to store the information we will generate

for agla in agla_to_tcas: #loops through each agla gene and checks the reciprical tcas gene
    if agla not in agla_to_ldec: #checks if agla has a reciprical ldec gene
        continue

    tcas = agla_to_tcas[agla] #identifies the tcas gene that correlates to agla
    ldec = agla_to_ldec[agla] #identifies the ldec gene that correlates to agla

    # Check that Tcas and Ldec are also reciprocal best hits
    if tcas in tcas_to_ldec and tcas_to_ldec[tcas] == ldec: #checks if tcas and ledc are reciprical
        three_way.append((agla, tcas, ldec)) #saves the information on agla, tcas, and ldec


print("Number of 3-way orthologous groups:", len(three_way))
```
